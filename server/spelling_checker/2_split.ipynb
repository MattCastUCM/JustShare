```python
import os
from loader import export_sentences, read_sentences
import math
import random
```

```python
language = "es"

data_dir = "./data"

sentences_path = os.path.join(data_dir, f"{language}_sentences.txt")
train_sentences_path = os.path.join(data_dir, f"{language}_train_sentences.txt")
test_sentences_path = os.path.join(data_dir, f"{language}_test_sentences.txt")

```

```python
def train_test_split(sentences: list[list[str]], train_ratio: float):
	sentences = sentences.copy()
	random.shuffle(sentences)

	split = math.ceil(train_ratio * len(sentences))

	train = sentences[:split]
	test = sentences[split:]

	return train, test
```

```python
sentences = read_sentences(sentences_path)
```

```python
train_sentences, test_sentences = train_test_split(sentences, train_ratio=0.9)

export_sentences(train_sentences_path, train_sentences)
export_sentences(test_sentences_path, test_sentences)
```
