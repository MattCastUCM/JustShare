```python
import pandas as pd
import os
```

```python
data_dir = "./data"

processed_dir = os.path.join(data_dir, "processed")
postest_path = os.path.join(processed_dir, "postest.csv")
```

```python
postest_df = pd.read_csv(postest_path, encoding="utf-8")
```

```python
print(f"LIKED MOST (total: {postest_df['liked_most'].dropna().shape[0]})")
for respuesta in postest_df["liked_most"].dropna():
    print(f"- {respuesta}")


print(f"\nCOMMENTS (total: {postest_df['comments'].dropna().shape[0]})")
for respuesta in postest_df["comments"].dropna():
    print(f"- {respuesta}")
```
